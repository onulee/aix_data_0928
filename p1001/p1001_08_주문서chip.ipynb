```python
# 1. item_name 당 주문서 개수를 출력하시오. 10개만 출력
item_count = df['item_name'].value_counts()[:10]
item_count
# item_count.index[0]
```

```python
df.groupby('item_name')['quantity'].sum()
```

```python
# 2. item_name 당 주문 총량(quantity)을 출력하시오.
item_quantity = df.groupby('item_name')['quantity'].sum()
item_quantity[:10] # item당 주문 총량을 출력
item_quantity.sort_values(ascending=False)
```

```python
# 3. 가장 많이 주문한 item_name  10개를 출력하시오.
df['item_name'].value_counts()[:10]
```

```python
df['item_name'].unique()[:10]  # 주문상품 종류
```

```python
# df['item_price'] = df['item_price'].str.replace('$','').str.strip().astype(float)
df['item_price'] = df['item_price'].apply(lambda x: float(x[1:]))
df.describe()
```

```python
# 주문당 평균 계산금액을 출력
df.groupby('order_id')['item_price'].sum().mean()
```

```python
# 한 주문에 10달러 이상 사용한 index.
chipo_orderid_group = df.groupby('order_id').sum()
results = chipo_orderid_group[chipo_orderid_group.item_price >= 10]

results.index.values
results[:10]
```
